```python
def v_c(t):
 vowel="aeiouAEIOU"
 v_count=0
 c_count=0
 for ch in t:
  if ch.isalpha():
   if ch in vowel:
     v_count+=1
   else:
     c_count+=1
 return v_count,c_count
res=input("Enterstring:")
vowels,consonants=v_c(res)
print("No.of vowels:",vowels)
print("No.of consonants:",consonants)
```

```python
def multi(mat_a,mat_b):
    if len(mat_a[0])!=len(mat_b):
        return None
    res=[]
    for i in range(len(mat_a)):
        row=[]
        for j in range(len(mat_b[0])):
            total=0
            for k in range(len(mat_b)):
                total+=mat_a[i][k]*mat_b[k][j]
            row.append(total)
        res.append(row)
    return res
r_a=int(input("enter rows of A:"))
c_b=int(input("enter cols of A:"))
mat_a=[]
print("ENter A:")
for i in range(r_a):
    row=list(map(int,input().split()))
    mat_a.append(row)
r_b=int(input("enter rows of B:"))
c_b=int(input("enter cols of B:"))
mat_b=[]
print("ENter B:")
for i in range(r_b):
    row=list(map(int,input().split()))
    mat_b.append(row)
res=multi(mat_a,mat_b)
if res is None:
    print("Error")
else:
    for row in res:
        print(row)
```

```python
def find_common(l1,l2):
    common=[]
    for i in l1:
        if i in l2 and i not in common:
            common.append(i)
    return common
l1=list(map(int,input("Enter 1st list elements:").split()))
l2=list(map(int,input("Enter 2nd list eleme:").split()))
res=find_common(l1,l2)
print("Common elements:",res)
```

```python
def transp(mat):
    transp=[]
    for i in range(len(mat[0])):
        row=[]
        for j in range(len(mat)):
            row.append(mat[j][i])
        transp.append(row)
    return transp
r=int(input("enter no.of rows:"))
c=int(input("enter no.of cols:"))
mat=[]
print("ENter matrix:")
for i in range(r):
    row=list(map(int,input().split()))
    mat.append(row)
res=transp(mat)
print("Transpose matrix:")
for row in res:
    print(row)
```

```python
import random
import statistics
def gen():
    num=[]
    for i in range(100):
        num.append(random.randint(100,150))
    mean=statistics.mean(num)
    median=statistics.median(num)
    mode=statistics.mode(num)
    return num,mean,median,mode
num,mean,median,mode=gen()
print("numbers generated:")
print(num)
print("mean:",mean)
print("median:",median)
print("mode:",mode)
```
