```python
import pandas as pd
import numpy as np
import time
import matplotlib.pyplot as plt
a=pd.read_csv(r"C:\Users\pshar\Downloads\Lab Session Data(IRCTC Stock Price).csv")
```

```python
m=np.mean(a["Price"])
v=np.var(a["Price"])
print("mean:",m)
print("var:",v)
```

```python
def mean(data):
    sum = 0
    for x in data:
        sum = sum + x
    return sum / len(data)
def var(data):
    m = mean(data)
    sum = 0
    for x in data:
        sum = sum + (x - m) ** 2
    return sum / len(data)
price = a["Price"].dropna() 
print("Mean:",mean(price))
print("Variance:",var(price))
```

```python
t=0
t1=0
for i in range(10):
    s=time.time()
    np.mean(a["Price"])
    np.var(a["Price"])
    t+=time.time()-s
    s=time.time()
    mean(a["Price"])
    var(a["Price"])
    t1+=time.time()-s
print("avg numpy time:",t/10)
print("avg fnc time:",t1/10)
```

```python
f11=f10=f01=f00=0
for x,y in zip(b1,b2):
    if x==1 and y==1:
        f11+=1
    elif x==1 and y==0:
        f10+=1
    elif x==0 and y==1:
        f01+=1
    else:
        f00+=1
JC=f11/(f11+f10+f01)
SMC=(f11+f00)/(f11+f10+f01+f00)
print("JC =",JC)
print("SMC =",SMC)
```
